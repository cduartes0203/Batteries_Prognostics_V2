```python
from pathlib import Path
import shutil
```

```python
# Caminhos principais
base_dir = Path(r"C:\Users\Usuario\Desktop\Python\Batteries_Prognostics\OLD\Opt")
dest_dir = base_dir / "Optimization"
dest_dir.mkdir(exist_ok=True)

# Busca todas as subpastas (ignorando a própria pasta 'Optimization')
subpastas = sorted([
    p for p in base_dir.iterdir() 
    if p.is_dir() and p.name != "Optimization"
])

contador = 1

for pasta in subpastas:
    arquivo_db = pasta / "optuna.db"
    
    # Verifica se o optuna.db existe dentro da subpasta
    if arquivo_db.exists():
        novo_nome = f"optuna_{contador:02d}.db"
        destino = dest_dir / novo_nome
        
        shutil.move(str(arquivo_db), str(destino))
        print(f"Movido: {pasta.name}\\optuna.db -> Optimization\\{novo_nome}")
        
        contador += 1

print(f"\nConcluído! {contador - 1} arquivo(s) processado(s).")
```
